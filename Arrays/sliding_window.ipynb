### What is Sliding Window?

A window that moves over the array

**Interview line:**

“Sliding window is an optimization technique used to reduce repeated calculations over subarrays.”

#### When Should You Use Sliding Window?

- Subarray / substring problems
- Continuous elements
- Sum, max, min, average
- Fixed or dynamic window size

- Not for scattered elements

```python
# Example 1:
# Find maximum sum of subarray of size k

arr = [2, 1, 5, 1, 3, 2]
k = 3

window_sum = sum(arr[:k])
max_sum = window_sum

for i in range(k, len(arr)):
    # Formula: window_sum += arr[i] - arr[i - k]

    window_sum += arr[i] # add next element
    window_sum -= arr[i - k] # subract pervious element
    max_sum = max(max_sum, window_sum)

print(max_sum)

# Time: O(n)
# Space: O(1)
```
