### What is Two Pointers?
Two pointers = two indexes moving through the array

**Interview line:**

“Two pointers is an optimization technique used to reduce time complexity by avoiding nested loops.”

#### When Should You Use Two Pointers?

- Array or string
- Sorted data (most of the time)
- Pair / reverse / compare problems
- Need O(n) instead of O(n²)

Not useful when random jumping is required

```python
# Example 1: Reverse an Array
# Bad Approach

arr = [1, 2, 3, 4]
rev = arr[::-1]
rev
```

```python
# Best Approach
arr = [1, 2, 3, 4]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1

print(arr)
```

```python
# Example 2: Check if Array is Palindrome

arr = [1, 2, 3, 4, 5, 4, 3, 2, 1]

left = 0
right = len(arr) - 1

is_palindrome = True

while left < right:
    if arr[left] != arr[right]:
        is_palindrome = False
        break

    left += 1
    right -= 1

print(is_palindrome) 

# Time: O(n)
# Space: O(1)
```

```python
# Example 3: Pair Sum

arr = [1, 2, 3, 4, 6]
target = 6

left = 0
right = len(arr) - 1

while left < right:
    curr_sum = arr[left] + arr[right]

    if curr_sum == target:
        print(arr[left], arr[right])
        break

    elif curr_sum < target:
        left += 1
    else:
        right -= 1

# Time: O(n)
# Space: O(1)
```

**Interview explanation:**

“Since the array is sorted, we move pointers based on the sum.”

### Pointer Movement Rule (VERY IMPORTANT)

Sum too small - left += 1
Sum too big   - right -= 1
Match found	  - Stop

**Memory trick:**
- Need bigger sum? Move left
- Need smaller sum? Move right

**Interview Traps**
- Forgetting array must be sorted
- Using two pointers when hash map is better
- Infinite loop (left <= right mistake)
